```sql
select top (3) *
from sql_recap.dbo.customers as c
join sql_recap.dbo.orders as o on c.customer_id = o.customer_id
```

```sql
select top (3) country, sum(total_amount) as sum_orders
from sql_recap.dbo.customers as c
join sql_recap.dbo.orders as o on c.customer_id = o.customer_id
group by country
order by sum_orders DESC
```

```sql
select top (5) country, sum(total_amount) as sum_orders
from sql_recap.dbo.customers as c
join sql_recap.dbo.orders as o on c.customer_id = o.customer_id
group by country
order by sum_orders DESC
```

```sql
select top (3) order_id , sum(total_amount) as sum_orders
from sql_recap.dbo.orders 
group by order_id
order by sum_orders
```

```sql
select order_id , sum(total_amount) as sum_orders
from sql_recap.dbo.orders 
group by order_id
having order_id = '2600'
order by sum_orders
```
