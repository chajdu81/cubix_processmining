```sql
SELECT top (3)
    order_id,
    order_date,
    total_amount,
    LAG(total_amount) over (order by order_date) as previous_order_amount
FROM 
    sql_recap.dbo.orders
ORDER BY
    order_date;
```

```sql
SELECT top (3)
    order_id,
    order_date,
    total_amount,
    LEAD(total_amount) over (order by order_date) as previous_order_amount
FROM 
    sql_recap.dbo.orders
ORDER BY
    order_date;
```

```sql
SELECT top (3)
    order_id,
    order_date,
    total_amount,
    RANK()  over (order by total_amount DESC) as order_rank
FROM 
    sql_recap.dbo.orders
ORDER BY
    order_rank;
```

```sql
IF OBJECT_ID('tempdb..#orders_ranked') IS NOT NULL
    DROP TABLE #orders_ranked;

SELECT
    order_id,
    total_amount,
    RANK() over (order by total_amount DESC) as order_rank
INTO
    #orders_ranked --temporary table using # as prefix
FROM 
    sql_recap.dbo.orders


SELECT
    order_id,
    total_amount,
    order_rank
FROM #orders_ranked
WHERE order_id = '2600' 
order by total_amount DESC


```
