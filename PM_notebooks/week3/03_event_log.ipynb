Most pedig készítsük el az event logot. A Microsoft Power Automate egy táblából használja az event logot, ez azt jelenti, hogy a Case és Event attribútumokat egy közös táblába kell beépíteni. Folytatva az előző két leckében használt lekérdezéseket:

```sql
--A case táblánk:
select o.*, i.tetelek_szama, i.teljes_darabszam from sql_recap.dbo.orders o
left join (select order_id, count(product_id) as tetelek_szama, sum(quantity) as teljes_darabszam from [sql_recap].[dbo].[order_details]
group by order_id) i 
on o.order_id=i.order_id
```

```sql
--Az Event tábláink:
-- Create order
Select order_id as CaseID, 'Create Order' as Event_type, order_date as Event_time, customer_id as 'User' from [sql_recap].[dbo].[orders]
```

```sql
--Customer sign-up
select o.order_id as CaseID, 'Customer Sign-up' as Event_type, signup_date as Event_time, c.customer_id as 'User' from [sql_recap].[dbo].[orders] o 
left join [sql_recap].[dbo].[customers] c
on o.customer_id=c.customer_id
```

Ahhoz, hogy az event logot össze tudjuk állítani, az event szekcióban összeállított lekérdezéseket fogjuk feldúsítani a case szekcióban összeállítótt lekérdezésből a szükséges adatokkal. Mivel az event log összefűzése általában több event táblából történik, fontos, hogy a különböző event táblák megegyező struktúrájúak legyenek - vagyis ugyanazok az oszlopok, ugyanolyan adattípússal szerepeljenek mindegyikben.

```sql
--Nézzük a customer sign-up eventet:
select o.order_id as CaseID, 'Customer Sign-up' as Event_type, signup_date as Event_time, c.customer_id as 'User', 
c.customer_id, c.country, i.tetelek_szama, i.teljes_darabszam, total_amount 
from sql_recap.dbo.orders o
left join (select order_id, count(product_id) as tetelek_szama, sum(quantity) as teljes_darabszam from [sql_recap].[dbo].[order_details]
group by order_id) i 
on o.order_id=i.order_id
left join [sql_recap].[dbo].[customers] c
on o.customer_id=c.customer_id
```

```sql
--Nézzük meg ugyanezt a create order eseményre:
Select o.order_id as CaseID, 'Create Order' as Event_type, order_date as Event_time, c.customer_id as 'User',  
c.customer_id, c.country, i.tetelek_szama, i.teljes_darabszam, total_amount 
from [sql_recap].[dbo].[orders] o
left join (select order_id, count(product_id) as tetelek_szama, sum(quantity) as teljes_darabszam from [sql_recap].[dbo].[order_details] group by order_id) i 
on o.order_id=i.order_id
left join [sql_recap].[dbo].[customers] c
on o.customer_id=c.customer_id
```

Most pedig fűzzük össze a kettőt, valamint állítsuk őket sorrendbe a Case ID és az Event time alapján

```sql
select * from 

(
select o.order_id as CaseID, 'Customer Sign-up' as Event_type, signup_date as Event_time, c.customer_id as 'User', 
c.customer_id, c.country, i.tetelek_szama, i.teljes_darabszam, total_amount from sql_recap.dbo.orders o

left join (select order_id, count(product_id) as tetelek_szama, sum(quantity) as teljes_darabszam from [sql_recap].[dbo].[order_details]
group by order_id) i 
on o.order_id=i.order_id

left join [sql_recap].[dbo].[customers] c
on o.customer_id=c.customer_id
--
union all
--
Select o.order_id as CaseID, 'Create Order' as Event_type, order_date as Event_time, c.customer_id as 'User',  
c.customer_id, c.country, i.tetelek_szama, i.teljes_darabszam, total_amount from [sql_recap].[dbo].[orders] o

left join (select order_id, count(product_id) as tetelek_szama, sum(quantity) as teljes_darabszam from [sql_recap].[dbo].[order_details]
group by order_id) i 
on o.order_id=i.order_id

left join [sql_recap].[dbo].[customers] c
on o.customer_id=c.customer_id) 

as Event_log

order by CaseID, Event_time
```
